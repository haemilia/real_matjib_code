```python
import duckdb
from pathlib import Path
```

# 네이버 데이터 DB로 옮기기

```python
# 네이버 데이터 파일 위치
datafiles_dir = Path("G:/My Drive/Data/naver_search_results")
restaurants_path = datafiles_dir / "restaurants_table.parquet"
navermap_reviews_path = datafiles_dir / "navermap_reviews.parquet"
naverblog_reviews_path = datafiles_dir / "naverblog_reviews_store_id.parquet" # store_id 추가 버전
navermap_reviews_labelled_path = datafiles_dir / "navermap_reviews_labelled.parquet" # 라벨링을 위해 샘플링한 애들만 모아둔 파일. 완전한 라벨링은 안 된 상태 but 라벨링 된 애들만 가지고 모델 훈련 중...
```

```python
# 새로운 DB 파일 위치 (G 드라이브, H 드라이브 등 어디에 있는 지 확인해주세요)
new_db_path = Path(r"H:\My Drive") / "reviews.db"
```

```python
with duckdb.connect(database=new_db_path, read_only=False) as db_conn:
    q1 = f"CREATE OR REPLACE TABLE naver_restaurants AS SELECT * FROM '{restaurants_path.as_posix()}'"
    db_conn.execute(q1)
    q2 = f"CREATE OR REPLACE TABLE navermap_reviews AS SELECT * FROM '{navermap_reviews_path.as_posix()}'"
    db_conn.execute(q2)
    q3 = f"CREATE OR REPLACE TABLE naverblog_reviews AS SELECT * FROM '{naverblog_reviews_path.as_posix()}'"
    db_conn.execute(q3)
    q4 = f"CREATE OR REPLACE TABLE navermap_reviews_labelled AS SELECT * FROM '{navermap_reviews_labelled_path.as_posix()}'"
    db_conn.execute(q4)
```

# DB에서 정보 보기

```python
with duckdb.connect(database=new_db_path, read_only=False) as db_conn:
    # DB에 어떤 테이블 있는 지 보기
    db_conn.sql("PRAGMA show_tables_expanded;").show() 
    # 데이터프레임으로 테이블 가져오기
    restaurants = db_conn.table("restaurants").to_df() 
    display(restaurants)
```

```python
with duckdb.connect(database=new_db_path, read_only=True) as db_conn:
    blog = db_conn.table("naverblog_reviews").to_df() 
    display(blog)
```
